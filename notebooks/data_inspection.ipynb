```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
print("Libraries imported successfully!")
```

```python
# Define project paths
PROJECT_DIR = Path.cwd().parent
DATA_DIR = PROJECT_DIR / "datasets" / "original"

# Load datasets
products = pd.read_csv(DATA_DIR / "products.csv")
locations = pd.read_csv(DATA_DIR / "warehouse_locations.csv")
order_lines = pd.read_csv(DATA_DIR / "order_lines.csv")
inventory = pd.read_csv(DATA_DIR / "inventory.csv")

print("All four datasets loaded successfully!")
```

```python
# Display dataset dimensions

print("Products:", products.shape)
print("Warehouse Locations:", locations.shape)
print("Order Lines:", order_lines.shape)
print("Inventory:", inventory.shape)
```

```python
# Display first five records from each dataset

display(products.head())
display(locations.head())
display(order_lines.head())
display(inventory.head())
```

```python
# Check column names in all datasets

print("PRODUCTS COLUMNS:")
print(products.columns.tolist())

print("\nWAREHOUSE LOCATIONS COLUMNS:")
print(locations.columns.tolist())

print("\nORDER LINES COLUMNS:")
print(order_lines.columns.tolist())

print("\nINVENTORY COLUMNS:")
print(inventory.columns.tolist())
```

```python
# Check data types of all datasets

print("PRODUCTS DATA TYPES:")
print(products.dtypes)

print("\nWAREHOUSE LOCATIONS DATA TYPES:")
print(locations.dtypes)

print("\nORDER LINES DATA TYPES:")
print(order_lines.dtypes)

print("\nINVENTORY DATA TYPES:")
print(inventory.dtypes)
```

```python
# Check missing values in all datasets

print("PRODUCTS MISSING VALUES:")
print(products.isnull().sum())

print("\nWAREHOUSE LOCATIONS MISSING VALUES:")
print(locations.isnull().sum())

print("\nORDER LINES MISSING VALUES:")
print(order_lines.isnull().sum())

print("\nINVENTORY MISSING VALUES:")
print(inventory.isnull().sum())
```

```python
# Total missing values

print("Products:", products.isnull().sum().sum())
print("Locations:", locations.isnull().sum().sum())
print("Order Lines:", order_lines.isnull().sum().sum())
print("Inventory:", inventory.isnull().sum().sum())
```

```python
# Check duplicate rows in all datasets

print("Products duplicates:", products.duplicated().sum())

print("Locations duplicates:", locations.duplicated().sum())

print("Order Lines duplicates:", order_lines.duplicated().sum())

print("Inventory duplicates:", inventory.duplicated().sum())
```

```python
# Display duplicate records

display(products[products.duplicated()])

display(locations[locations.duplicated()])

display(order_lines[order_lines.duplicated()])

display(inventory[inventory.duplicated()])
```

```python
# Check unique IDs
print("PRODUCTS")
print("Total SKUs:", len(products))
print("Unique SKUs:", products["SKU_ID"].nunique())

print("\nWAREHOUSE LOCATIONS")
print("Total Locations:", len(locations))
print("Unique Locations:", locations["Location_ID"].nunique())

print("\nORDER LINES")
print("Total Order Lines:", len(order_lines))
print("Unique Order Lines:", order_lines["Order_Line_ID"].nunique())

print("\nINVENTORY")
print("Total Inventory Records:", len(inventory))
print("Unique SKUs:", inventory["SKU_ID"].nunique())
```
